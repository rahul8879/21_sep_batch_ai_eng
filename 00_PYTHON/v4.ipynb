```python

data['session_date'] = pd.to_datetime(data['session_date']).dt.strftime('%d-%m-%Y')
```

```python
data['session_date'] = pd.to_datetime(data['session_date'])
data['feedback_date'] = pd.to_datetime(data['feedback_date'])
```

```python
data['month'] = data['session_date'].dt.month
data['year'] = data['session_date'].dt.year
data['day'] = data['session_date'].dt.day
data['week'] = data['session_date'].dt.isocalendar().week
```
