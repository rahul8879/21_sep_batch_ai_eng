```python
import pandas as pd
data = pd.read_csv("data/bank_data.csv")
data.head(2)

data.shape
```

```python
data.head(2) # select top 2 * from data
```

```python
data.info()
```

```python
data.status.value_counts()
```

```python
def category(x):
    if x =='Rejected':
        return 0
    else:
        return 1

def formate_income(x):
    return int(x.split('₹')[1].replace(',',''))
```

```python
data['status'] = data['status'].apply(category)
data['income'] = data['income'].apply(formate_income)
```

```python
data.head(5)
```

```python
# int('₹35,000'.split('₹')[1].replace(',',''))
data.info()
```

```python
X = data[['cibil_score','income']]
y = data['status']
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,y, test_size=0.1)
```

```python
x_train.shape,y_train.shape
x_test.shape,y_test.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier

# step 1 : setting up all the param --> if you dont ---> default param
knn = KNeighborsClassifier(n_neighbors=8)


# ste2 : training the model
knn.fit(x_train,y_train)
```

```python
x_test.head()
```

```python
# test the model
test = [850,25000]
knn.predict(x_test)
```

```python
test_df['actual'] = y_test
test_df['predicted'] = knn.predict(x_test)

test_df
```
