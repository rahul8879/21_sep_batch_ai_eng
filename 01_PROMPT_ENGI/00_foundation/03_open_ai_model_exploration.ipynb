```python
from dotenv import load_dotenv
from openai import OpenAI
load_dotenv()  # Load environment variables from .env file
```

```python
client = OpenAI()
response = client.chat.completions.create(
    model = "gpt-4o",
    messages=[
        {"role": "user", "content": "what is 2+2"}
    ]
)
```

```python
def classifiy_email(prompt):
    response = client.chat.completions.create(
        model = "gpt-4o",
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content
```

```python
email = "Hi team I am facing some issue with my account"

prompt = f"Classify the following email into categories and just give me category: {email}"

classifiy_email(prompt)
```

```python
prompt_v2 = f"""You are an expert support email classifier. 
Classify the email into exactly one of: 
[Billing, Technical, Feature Request, Spam, Other]. 
Return only the category name,
nothing else. Email: {email}"""


classifiy_email(prompt_v2)
```
