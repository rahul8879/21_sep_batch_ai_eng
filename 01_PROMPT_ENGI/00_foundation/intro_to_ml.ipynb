```python
import pandas as pd

data = {
    'income':[5, 7, 9, 2, 4], # between 0-10
    'cibil_score':[0.72, 0.78, 0.83, 0.67, 0.65], # between 0-10
    'approved':[1, 1, 1, 0, 0] # 1: Yes, 0: No

}

df = pd.DataFrame(data)
df
```

```python
# v1 = [3,2]
# v2 = [3,2]

import math
def calculate_distance(df, new_customer, top_k=3):
    distances = []
    for index, row in df.iterrows():
        distance = math.sqrt((row['income'] - new_customer[0])**2 + (row['cibil_score'] - new_customer[1])**2)
        distances.append(round(distance, 2))

    # sort it and return the majority class
    df['distance'] = distances
    print(distances)
    return df.sort_values('distance').iloc[:top_k]['approved']


# time complexity 
# o(n*log(n))
```

```python
calculate_distance(df, [4,0.9])
```
