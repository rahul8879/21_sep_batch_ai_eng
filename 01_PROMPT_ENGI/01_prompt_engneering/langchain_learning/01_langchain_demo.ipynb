```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
# lets import prompt template class
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()
# 
model = ChatOpenAI(model='gpt-6-astra')
```

```python
# I will call it 

response = model.invoke('hello')
```

```python
response.content
```

```python
# step 1 : we defined our template

template = ChatPromptTemplate.from_template(
    'Reply in one sentence : {question}'
)


# step 2 : Define the flow/chain
chain = template | model
```

```python
# step 3 : just pass the input and get the reponse 

chain.invoke('what was my last questions')
```

```python
# eg 2: 

# step 1 : 
classify_prompt = ChatPromptTemplate.from_template(
    'Classify this email in ONE word (Billing/Technical/Spam/Other):\n'
    'Subject: {subject}\nBody: {body}'
)

# step 2 :

classify_chain = classify_prompt | model | StrOutputParser()

# step 3 : lets pass the input

classify_chain.invoke({'subject': 'Payment failed',   'body': 'Charged twice this month.'})
```

```python
emails = [
    {'subject': 'Payment failed',   'body': 'Charged twice this month.'},
    {'subject': 'App crash',        'body': 'Crashes on Settings page.'},
    {'subject': 'Add Slack?',       'body': 'Want Slack notifications.'},
    {'subject': 'WIN FREE IPHONE',  'body': 'Click here NOW!!!'},
    {'subject': 'Invoice question', 'body': 'Can I get annual invoice?'},
]

result = []

for i in emails:
    output = classify_chain.invoke(i)
    result.append(output)
```

```python
result
```

```python
# invoke/batch/stream

response = classify_chain.batch(emails,config={'max_concurrency':3})
```

```python
response
```

```python
# any use case

# step 1 : define your prompts
# step 2 : create your chain

# step 3: call you chain ----> if you have only one records ---> invoke() ---> but if you want to process n number records parallaly then 
# --> batch
```
