```python
# manual stiching 

user_turn_1 = "I received a billing complaint from John about invoice #1042. He was charged $500 instead of $250."
ai_turn_1 = llm.invoke(user_turn_1)

print("TURN 1")
print(f"User: {user_turn_1}")
print(f"AI:   {ai_turn_1}")
print()

```

```python
user_turn_2 = "What category does his complaint fall under?"


stitched_prompt_turn2 = f"""\
Previous conversation:
User: {user_turn_1}
AI: {ai_turn_1}

Now answer:
User: {user_turn_2}"""


ai_turn_2 = llm.invoke(stitched_prompt_turn2)
print("TURN 2")
print(f"User: {user_turn_2}")
print(f"AI:   {ai_turn_2}")
print()
```

```python
message = []

def add_message(role, content):
    message.append({"role": role, "content": content})

questions = ' my name is rahul'
ans = llm.invoke(questions)
add_message("user", questions)
add_message("ai", ans)

```

```python
# follow up questions
questions = 'what is my name'


stitched_prompt_turn2 = f"""\
Previous conversation:
{message}

Now answer:
User: {questions}"""

ans = llm.invoke(stitched_prompt_turn2)

add_message("user", questions)
add_message("ai", ans)
```

```python
message = []


def add_message(role, content):
    message.append({"role": role, "content": content})


max_iter = 10
step = 0
while max_iter >step:
    questions = input("User: ")
    stitched_prompt = f"""\
    Previous conversation:
    {message}

    Now answer:
    User: {questions}"""

    ans = llm.invoke(stitched_prompt)
    print('Ans from AI:', ans)
    add_message("user", questions)
    add_message("ai", ans)
    step +=1
```
